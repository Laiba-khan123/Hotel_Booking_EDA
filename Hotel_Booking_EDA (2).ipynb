```python
# Step 1: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# For better visuals
sns.set(style="whitegrid")
%matplotlib inline
```

```python
# Step 3: Load the Dataset
file_path = '/content/drive/MyDrive/hotel_booking.csv'
df = pd.read_csv(file_path)
```

```python
# Step 4: Initial Data Inspection
# First few rows
df.head()
```

```python
# Summary of the dataset
df.info()
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Statistical summary
df.describe()
```

```python
# Shape of the dataset
print("Shape:", df.shape)
```

```python
 # Step 5: Data Cleaning
# Drop columns with excessive missing values or fill if appropriate
df['children'].fillna(0, inplace=True)
df['country'].fillna(df['country'].mode()[0], inplace=True)
df['agent'].fillna(0, inplace=True)
df['company'].fillna(0, inplace=True)
```

```python
# Convert dates
df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'])
```

```python
# Fix data types
df['agent'] = df['agent'].astype(int)
df['company'] = df['company'].astype(int)
```

```python
# step 6: Feature Engineering
df['total_guests'] = df['adults'] + df['children'] + df['babies']
df['stay_length'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']
df['is_family'] = (df['children'] > 0) | (df['babies'] > 0)
```

```python
# Step 7 : EDA (Detailed Analysis)
# Booking Status Overview
sns.countplot(data=df, x='is_canceled')
plt.title("Booking Cancellation Count")
plt.xticks([0, 1], ['Not Canceled', 'Canceled'])
plt.show()

# Percentage
df['is_canceled'].value_counts(normalize=True) * 100
```

```python
# 6.2 Booking by Hotel Type
sns.countplot(data=df, x='hotel', hue='is_canceled')
plt.title("Booking and Cancellation by Hotel Type")
```

```python
# 6.3 Seasonal Trends
# Monthly trends
monthly = df.groupby(['arrival_date_month'])['is_canceled'].mean().sort_index()
monthly.plot(kind='bar')
plt.title("Average Cancellation Rate per Month")
```

```python
# 6.4 Lead Time Effect
sns.histplot(df[df['is_canceled']==1]['lead_time'], color='red', label='Canceled', kde=True)
sns.histplot(df[df['is_canceled']==0]['lead_time'], color='green', label='Not Canceled', kde=True)
plt.legend()
plt.title("Lead Time Distribution")

```

```python
# 6.5 Distribution of Stay Duration
sns.boxplot(data=df, x='is_canceled', y='stay_length')
plt.title("Stay Duration vs Cancellation")

```

```python
# 6.6 Country-wise Booking Analysis
top_countries = df['country'].value_counts().head(10)
sns.barplot(x=top_countries.values, y=top_countries.index)
plt.title("Top Countries by Booking Count")
```

```python
# 6.7 Market Segment & Distribution Channel
sns.countplot(data=df, y='market_segment', order=df['market_segment'].value_counts().index, hue='is_canceled')
plt.title("Market Segment and Cancellation Rate")

sns.countplot(data=df, y='distribution_channel', order=df['distribution_channel'].value_counts().index, hue='is_canceled')
plt.title("Distribution Channel and Cancellation Rate")
```

```python
# 6.8 Room Type Preferences
sns.countplot(data=df, x='reserved_room_type', hue='is_canceled')
plt.title("Room Preferences and Cancellations")
```

```python
# step 7 correlation HeatMap
plt.figure(figsize=(12, 8))
corr = df[['is_canceled', 'lead_time', 'total_guests', 'stay_length', 'adults', 'children', 'babies']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```
